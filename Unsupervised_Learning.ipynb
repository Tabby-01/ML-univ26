```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

np.random.seed(67)
```

```python
# X = np.vstack((np.random.rand(500, 2) + [2, 2],
#     np.random.rand(500, 2) + [2.5, 3],
#     np.random.rand(500, 2) + [3, 4],
#     np.random.rand(500, 2) + [3.5, 5]))
# print(X)

from sklearn.datasets import make_blobs

X, y_true = make_blobs(
    n_samples=1000,
    centers=[[2, 2], [2.5, 3], [3, 4], [3.5, 5]],
    cluster_std = 1.5,
    random_state=67
)
```

```python
plt.scatter(X[:,0], X[:,1], s=10, color='orange')
plt.title('Input')
plt.xlabel('X1')
plt.ylabel('X2')
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=67)
kmeans.fit(X)
```

```python
labels = kmeans.labels_
centroids = kmeans.cluster_centers_
```

```python
plt.scatter(X[:,0], X[:,1], c=labels, cmap='viridis', s=10)
plt.scatter(centroids[:,0], centroids[:,1], color='red', marker='x', s=70, label='Centroinds')
plt.title("Clusterisation Result")
plt.xlabel('X1')
plt.ylabel('X2')
plt.legend()
plt.show()
```
