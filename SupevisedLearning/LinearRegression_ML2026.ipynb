```python
import  numpy as np
import pandas as pd
```

```python

data = {
    "days": [1,2,3,4,5,6,7],
    "coins": [2,4,6,8,10,12,14]
}

df=pd.DataFrame(data)
df
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['days'], df['coins'], color='blue')
plt.title('title*')
plt.xlabel('days')
plt.ylabel('coins')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

X = df[['days']] #features - ознаки
y = df['coins'] #target!

model = LinearRegression()
model.fit(X, y)

print("Вага (W):", model.coef_)
print(" (b):", model.intercept_)
```

```python
y_pred = model.predict(X)

plt.scatter(X, y, color='blue')
plt.plot(X, y_pred, color="red")
plt.title('LinearRegression')
plt.xlabel('days')
plt.ylabel('coins')
plt.show()
```

```python
new_value = [[10]]
predicted_coins = model.predict(new_value)
print(f"Predict for {new_value[0][0]} days: {predicted_coins[0]} coins")
```
